# AI 패러다임의 발전과 주요 기술 분류
   - 인공지능(AI)은 인간의 인지·학습·판단 능력을 기계로 구현하는 기술 체계로, 구현 방식과 기술 발전 단계에 따라 구분됩니다.

   1) 규칙 기반 시스템 (Rule-based System)
      - 개념: 전문가가 설계한 정교한 조건문(If-Else) 논리에 따라 입력을 처리하는 방식입니다.
      - 특징:
        - 명시적인 예외 처리 및 결정론적 로직 구현에 유리합니다.
        - 데이터 학습 과정이 없으므로 조건 규칙이 복잡해질수록 유지보수 비용이 급증하는 한계가 있습니다.

   2) 머신러닝 (Machine Learning, ML)
      - 개념: 명시적인 프로그래밍 없이, 축적된 데이터를 통계적으로 분석하여 패턴을 자동 도출하는 기술입니다.
      - 특징:
        - 데이터 고유의 특징(Feature)을 기반으로 자율적인 추론 및 미래 예측 모델을 생성합니다.
        - 현대 AI 시스템 구축의 핵심 프레임워크 역할을 담당합니다.

   3) 딥러닝 (Deep Learning, DL)
      - 개념: 머신러닝의 하위 분야로, 생물학적 뇌 구조를 모방한 다층 인공신경망(Multi-layer ANN) 알고리즘입니다.
      - 특징:
        - 데이터 특징을 사람이 직접 정제하지 않아도 모델 내부에서 자율적으로 추출(Feature Extraction)합니다.
        - 고성능 GPU 연산을 기반으로 이미지, 음성, 텍스트 등 고차원 비정형 데이터 분석에 탁월한 성과를 보입니다.

## 차세대 인공지능 기술 트렌드
   - AI는 빠르게 진화하고 있으며, 다양한 형태의 데이터와 상황에 대응하는 기술들이 등장하고 있습니다.
   - 대표적인 핵심 기술 트렌드는 다음과 같습니다.

   1) 멀티모달 AI (Multimodal AI)
      - 개념: 텍스트뿐만 아니라 이미지, 음성, 영상, 코드 등 여러 종류의 모달리티 데이터를 동시에 이해하고 처리하는 기술입니다.
      - 특징:
        - 단일 형태 데이터에 국한되지 않고 복합적인 맥락을 파악합니다.
        - 시각 분석, 음성 대화, 문서 이해가 결합된 고도화된 서비스를 구현합니다.

   2) 에이전트형 AI (AI Agent)
      - 개념: 단순 응답을 넘어 주어진 목표를 달성하기 위해 스스로 계획을 수립하고 도구를 활용하여 작업을 수행하는 자율형 시스템입니다.
      - 특징:
        - 목표 분해, 외부 API 호출, 연동 프로그램 실행 등을 자율적으로 처리합니다.
        - 업무 자동화 및 워크플로우 대행 영역에서 활용도가 높습니다.

   3) 검색 증강 생성 (Retrieval-Augmented Generation, RAG)
      - 개념: 외부 DB나 최신 문서에서 관련 정보를 검색한 후, 이를 기반으로 답변을 생성하는 기술입니다.
      - 특징:
        - 환각(Hallucination) 현상을 크게 줄이고 답변의 정확도를 높입니다.
        - 최신 정보 반영 및 기업 내부 데이터 기반 질의응답 시스템 구축에 필수적입니다.

   4) 온디바이스 AI (On-Device AI)
      - 개념: 중앙 서버나 클라우드를 거치지 않고 스마트폰, PC 등 기기 자체에서 직접 AI 모델을 구동하는 기술입니다.
      - 특징:
        - 인터넷 연결 없이 작동하며 반응 속도가 빠릅니다.
        - 데이터가 외부로 유출되지 않아 개인정보 보호 및 보안성이 우수합니다.

## 머신러닝의 주요 활용 분야 (Task)

   1) 분류 (Classification)
      - 개념: 입력 데이터를 사전에 정의된 몇 가지 카테고리(범주) 중 하나로 나누는 작업입니다.
      - 예시: 스팸 이메일 차단 시스템 (스팸 vs 정상 메일)

   2) 회귀 (Regression)
      - 개념: 데이터의 연관성을 분석해 연속적인 숫자 형태의 값을 예측하는 작업입니다.
      - 예시: 주변 시세와 면적 데이터를 기반으로 한 주택 가격 예측

   3) 클러스터링 (Clustering)
      - 개념: 데이터 자체의 유사성을 기반으로 별도의 정답 없이 비슷한 특성을 가진 데이터들을 하나의 그룹으로 묶는 작업입니다.
      - 예시: 구매 패턴에 따른 고객 세그먼트 분석

   4) 추천 시스템 (Recommendation System)
      - 개념: 사용자의 과거 행동 이력과 선호도 데이터를 분석하여 맞춤형 콘텐츠를 제안하는 기술입니다.
      - 예시: 넷플릭스나 유튜브의 영상 추천 알고리즘

   5) 이상 탐지 (Anomaly Detection)
      - 개념: 대다수의 정상적인 데이터 패턴에서 크게 벗어난 특이한 데이터나 이상치를 찾아내는 기법입니다.
      - 예시: 신용카드 부정 사용 및 금융 사기 거래 탐지

   6) 강화 학습 응용 (Reinforcement Learning Task)
      - 개념: 에이전트가 가상의 환경 안에서 시행착오를 겪으며 최적의 행동 노선을 찾아내도록 만드는 분야입니다.
      - 예시: 자율주행 자동차 구동, 게임 플레이, 바둑 인공지능 알파고

## 머신러닝의 다양한 학습 방법

   1) 지도 학습 (Supervised Learning)
      - 개념: 입력 데이터와 함께 정답에 해당하는 레이블(Label)이 제공되는 환경에서 학습합니다.
      - 특징: 주어진 입력에 대해 올바른 출력(레이블)을 예측하는 모델을 구축합니다.

   2) 비지도 학습 (Unsupervised Learning)
      - 개념: 레이블(정답) 없이 입력 데이터 자체의 구조나 패턴을 학습합니다.
      - 특징: 데이터 내부의 숨겨진 구조를 발견하거나 군집화를 수행합니다.

   3) 차원 축소 (Dimensionality Reduction)
      - 개념: 고차원 데이터의 정보를 최대한 보존하면서 중요한 특성 중심으로 차원을 줄이는 비지도 학습 기법입니다.
      - 특징: 데이터 시각화나 연산 효율성을 높이는 데 활용됩니다.

   4) 강화 학습 (Reinforcement Learning)
      - 개념: 에이전트(Agent)가 환경(Environment)과 상호작용하며 보상(Reward)을 최대화하는 방향으로 행동을 학습합니다.
      - 특징: 명시적인 정답 대신 보상을 통해 최적의 의사결정 전략을 도출합니다.

   5) 반지도 학습 (Semi-Supervised Learning)
      - 개념: 소량의 레이블 데이터와 대량의 미레이블 데이터를 결합하여 학습을 진행합니다.
      - 특징: 데이터 레이블링 비용이 많이 들고 구축이 어려울 때 효과적입니다.

   6) 자기 지도 학습 (Self-Supervised Learning)
      - 개념: 데이터 자체에서 정답을 생성하여 자율적으로 학습을 유도하는 방식입니다.
      - 특징: 데이터 일부를 마스킹(가림)하고 이를 맞추는 과정을 통해 대규모 데이터의 표현력을 사전 학습합니다.

   7) 온라인 학습 (Online Learning)
      - 개념: 전체 데이터를 일괄 학습시키지 않고, 데이터가 들어오는 대로 실시간으로 모델을 업데이트합니다.
      - 특징: 데이터 스트림이 실시간으로 변하거나 지속적인 적응이 필요한 환경에 적합합니다.

   8) 전이 학습 (Transfer Learning)
      - 개념: 대규모 데이터로 미리 학습된 모델(Pre-trained Model)의 지식을 새로운 유사 과제에 재활용하는 기법입니다.
      - 특징: 학습 시간을 대폭 단축하며, 소량의 데이터로도 높은 성능을 도출할 수 있습니다.

## 핵심 메커니즘: 모델과 학습의 본질

   1) 모델 (Model)
      - 개념: 데이터를 학습하여 패턴과 규칙을 찾아내고, 새로운 데이터에 대해 예측이나 결정을 내리는 수학적 구조 또는 알고리즘입니다.
      - 특징:
        - 입력 데이터를 처리하여 정답을 도출하는 시스템의 핵심 구성 요소입니다.
        - 수학적 함수 구조를 바탕으로 하며, 가중치(Weights)와 편향(Biases)이라는 매개변수로 구성됩니다.
        - 검증된 기존 모델 파이프라인을 목적에 맞게 재활용하고 최적화하는 방식이 효율적입니다.

   2) 학습 (Training)
      - 개념: 데이터로부터 오류를 줄여나가며 모델의 매개변수(가중치, 편향)를 최적화하는 과정입니다.
      - 특징:
        - 입력 데이터와 정답 간의 오차를 계산하고, 역전파(Backpropagation)와 최적화 알고리즘을 통해 매개변수를 반복 조정합니다.
        - 반복적인 오차 수정 과정을 거쳐 모델의 예측 정확도를 향상시킵니다.

   3) 학습 프로세스 (Training Process)
      - 메커니즘: 데이터 입력 -> 패턴 탐색 -> 수식화 -> 모델 구축 -> 실전 적용
      - 예시:
        - 평면 상의 데이터 점들을 가장 잘 표현하는 하나의 직선(수식)을 찾는 과정과 유사합니다.
        - 초기 직선의 오차를 바탕으로 기울기(가중치)와 Y절편(편향)을 미세하게 조정합니다.
        - 오차가 최소화되는 최적의 매개변수를 도출할 때까지 조정을 반복합니다.

## 핵심 기술 한눈에 보기

   | 구분 | 규칙 기반 (Rule-based) | 머신러닝 (Machine Learning) | 딥러닝 (Deep Learning) |
   | :--- | :--- | :--- | :--- |
   | 의사결정 주체 | 개발자 (사전 정의 규칙) | 알고리즘 (데이터 패턴 추출) | 다층 신경망 (고차원 특징 자율 학습) |
   | 주요 대상 데이터 | 정형 규칙 및 수식 | 정형·구조화 데이터 | 비정형 데이터 (이미지, 음성, 텍스트) |
   | 확장성 | 낮음 (규칙 추가의 한계) | 보통 (특징 공학 필요) | 높음 (대용량 데이터 수용) |
